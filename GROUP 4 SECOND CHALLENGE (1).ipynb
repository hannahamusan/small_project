 Find the Maximum and Minimum Values in a List Without Using max() and min() - 10 points
Given a tuple, find the maximum and minimum values without using the built-in max() or min() functions.
Example Input & Output

data = [8, 3, 9, 1, 6]
Output: Min: 1, Max: 9

```python
numbers = [2,3,4,9,20,15]
numbers.sort()
print(numbers)

print(numbers[0])
print(numbers[-1])

min_number=numbers[0]
max_number=numbers[-1]
print(f"The lowest number is {min_number}")
print(f" The highest number is {max_number}")
```
