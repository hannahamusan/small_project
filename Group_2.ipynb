## 2: Monthly Sales Report - 10 points
A store's sales for 4 weeks across 3 products are stored in this 2D array (rows = weeks, columns = products):.

sales = np.array([[120, 340, 210],
                   [150, 300, 250],
                   [170, 360, 230],
                   [140, 320, 260]])
Find the total sales per product and the total sales per week

```python
sales = np.array([[120, 340, 210], [150, 300, 250], [170, 360, 230], [140, 320, 260]])

print('Total sales per product')
product1 = sales[ : , 0]
total = sum(product1)
total
print(f' Product 1 total sales: {total}')

product2 = sales[ : , 1]
total = sum(product2)
total
print(f' Product 2 total sales: {total}')

product3 = sales[ : , 2]
total = sum(product3)
total
print(f' Product 3 total sales: {total}')

print('-' * 30)

print('Total sales per week')

week1 = sales[0, : ]
total = sum(week1)
print(f' Week 1 total sales: {total}')

week2 = sales[1, : ]
total = sum(week2)
print(f' Week 2 total sales: {total}')

week3 = sales[2, : ]
total = sum(week3)
print(f' Week 3 total sales: {total}')

week4 = sales[3, : ]
total = sum(week4)
print(f' Week 4 total sales: {total}')
```

## 5: Employee Bonus Filter - 10 points
An array holds employee performance scores out of 100: [45, 78, 92, 60, 88, 33, 95, 70]. To give a 20k bonus to employees who scored above 80, flag them as 1 and 0 otherwise. Also print how many employees qualify.

```python
score = np.array([45, 78, 92, 60, 88, 33, 95, 70])
score
```

```python
qualified = np.where(score > 80, 1, 0)
qualified
```

```python
np.sum(qualified)
```
