## QUESTION 1

 1: Weekly Temperature Tracker - 10 points
You recorded daily temperatures for a week: [22, 24, 19, 23, 25, 21, 20]. Find the average, the hottest day's temperature, the coldest day's temperature, and how many days were above 22.


```python
import numpy as np
```

```python
temperatures = np.array([22, 24, 19, 23, 25, 21, 20])
average_temp = np.mean(temperatures)
hottest_temp = np.max(temperatures)
coldest_temp = np.min(temperatures)
above_22 = np.sum(temperatures > 22)

print("Average Temperature :", average_temp)
print("Hottest Temperature:", hottest_temp)
print("Coldest Temperature:", coldest_temp)
print("Above 22:", above_22)
```

4: Exam Score Ranking - 10 points
An array holds exam scores for 10 students: [67, 88, 45, 92, 76, 59, 64, 30, 71, 51]. Sort the scores from lowest to highest, and find the index (position) of the student with the highest score in the original array.



```python
scores = np.array( [67, 88, 45, 92, 76, 59, 64, 30, 71, 51])
sorting = np.sort(scores)
position = np .argmax(scores)
print("Sorted List:", sorting)
print("Position:", position)
```

## QUESTION 5

5: Employee Bonus Filter - 10 points
An array holds employee performance scores out of 100: [45, 78, 92, 60, 88, 33, 95, 70]. To give a 20k bonus to employees who scored above 80, flag them as 1 and 0 otherwise. Also print how many employees qualify.``

```python
employees = np.array([45, 78, 92, 60, 88, 33, 95, 70])
above_80 = np.where(employees > 80, 1, 0)
qualified = np.sum(above_80)
print(above_80)
print(qualified)
```
